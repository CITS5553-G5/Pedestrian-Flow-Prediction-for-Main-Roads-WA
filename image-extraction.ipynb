```python
import requests

# 1. SETUP PARAMETERS
# Get your free token from the Mapillary Developer Dashboard
MAPILLARY_TOKEN = "YOUR_MAPILLARY_ACCESS_TOKEN"

# Target location coordinates
LATITUDE = 40.7580
LONGITUDE = -73.9855

# 2. SEARCH FOR CLOSEST IMAGES
# Search for images within a small bounding box radius around coordinates
bbox = f"{LONGITUDE-0.001},{LATITUDE-0.001},{LONGITUDE+0.001},{LATITUDE+0.001}"
search_url = (
    f"https://mapillary.com{bbox}&access_token={MAPILLARY_TOKEN}"
)

print("Searching for free street images...")
response = requests.get(search_url)

if response.status_code == 200:
    data = response.json()
    if data["data"]:
        # Get the ID of the first found image
        image_id = data["data"][0]["id"]
        print(f"Found image ID: {image_id}")

        # 3. FETCH THE ACTUAL IMAGE URL
        image_url = f"https://mapillary.com{image_id}?fields=thumb_1024_url&access_token={MAPILLARY_TOKEN}"
        img_response = requests.get(image_url).json()

        download_link = img_response.get("thumb_1024_url")
        print(f"Direct download link for your testing: {download_link}")
    else:
        print("No images found in this specific block radius.")
else:
    print(f"Error: {response.status_code}")
    print(response.text)
```
