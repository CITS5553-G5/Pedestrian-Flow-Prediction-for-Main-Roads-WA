# Microsoft Global ML Building Footprints — 49-site feature table

## What this data source is

[Microsoft Global ML Building Footprints](https://github.com/microsoft/GlobalMLBuildingFootprints) is an open global dataset of building-footprint polygons detected from aerial and satellite imagery by machine-learning models. A polygon represents the estimated ground outline of one detected building. The source is therefore useful for describing the **amount and concentration of built form around a survey site**; it is not an image collection and it is not a register of cadastral parcels or confirmed building uses.

### API and access method

This data source does **not** provide a REST, ArcGIS or coordinate-based query API. The official access method is file-based: the notebook sends standard HTTPS `GET` requests to Microsoft's public Azure Blob Storage, first to download the release's `dataset-links.csv` index and then to download the indexed `.csv.gz` partition files needed for the study sites. In other words, `requests.get()` is used as an HTTP file-download client; there is no API key, authentication token or API query endpoint.

Microsoft distributes the records as line-delimited GeoJSON files, divided by country and level-9 Bing quadkey. This notebook pins the **24 July 2026** release index, whose source documentation describes footprints derived from imagery acquired between 2014 and 2024. The data are licensed under [CDLA Permissive 2.0](https://cdla.dev/permissive-2-0/). The notebook downloads only the Australia partitions required by the 49 sites and caches the unchanged source files under `data/external_raw/`.

## What information can be obtained

The source directly supplies building polygon geometry. From that geometry, this notebook can calculate how many detected footprints intersect a defined area, how much ground area they occupy, and the proportion of the study area covered by detected buildings. These measures can act as simple indicators of surrounding development intensity. The polygons do **not** identify building type, use, occupancy, entrances, parking spaces, pedestrian access or population, so this notebook does not claim or estimate those attributes.

## Features produced by this notebook

For each `site_id`, the notebook measures detected buildings within an exact **500 m straight-line buffer** around the site coordinate and writes the following independent feature table:

| Output field | Plain meaning |
|---|---|
| `ms_building_count_500m` | Number of detected building footprints that overlap the 500 m buffer by a positive area. It describes the amount of separate detected building geometry around the site. |
| `ms_building_count_per_sqkm_500m` | The footprint count divided by the buffer area in square kilometres. It expresses detected building frequency on a common area basis. |
| `ms_building_footprint_area_m2_500m` | Total ground area, in square metres, covered by the union of detected building footprints inside the buffer. Parts outside the buffer are excluded and overlapping polygons are not double-counted. |
| `ms_building_coverage_ratio_500m` | Building-footprint area divided by total buffer area. For example, `0.10` means detected building polygons cover 10% of the 500 m buffer. |
| `ms_building_data_status` | Processing-status field showing whether calculation succeeded with buildings, succeeded with no detected buildings, or could not be completed because a required source partition was unavailable. It is a data-quality field, not an environmental measurement. |

All sites currently use the same 500 m buffer. Consequently, count and count per square kilometre contain the same relative information, while footprint area and coverage ratio contain the same relative information. Each pair should not be entered together as independent predictors in the same model.

## Scope and interpretation

This notebook reads the 49 authoritative sites from `data/processed/model_dataset.xlsx`, uses `site_id` as the sole join key, and outputs only the Microsoft status and features. Targets, QA fields and unrelated location attributes remain in `model_dataset.xlsx`. A detected footprint is not a confirmed cadastral building: completeness, positional accuracy and imagery date can vary by location. A zero count means that this release contains no detected footprint in the buffer; it does not prove that no building exists in reality.

# Microsoft 全球机器学习建筑轮廓——49 地点 feature 表

## 这个数据源是什么

[Microsoft Global ML Building Footprints](https://github.com/microsoft/GlobalMLBuildingFootprints) 是一个开放的全球建筑轮廓数据集。Microsoft 使用机器学习模型从航空和卫星影像中识别建筑，每个多边形表示一栋被识别建筑的大致地面轮廓。因此，它适合描述调查地点周边的**建筑数量和建成环境集中程度**；它不是影像库，也不是经过地籍确认的土地宗地或建筑用途登记。

### API 和接入方式

这个数据源**没有** REST API、ArcGIS API，也没有可以输入坐标后返回建筑记录的查询接口。官方接入方式是文件下载：Notebook 通过标准 HTTPS `GET` 请求访问 Microsoft 公共 Azure Blob Storage，先下载该版本的 `dataset-links.csv` 索引，再根据索引下载调查地点所需的 `.csv.gz` 数据分区。因此，代码中的 `requests.get()` 只是 HTTP 文件下载工具，不需要 API key、身份验证 token 或 API 查询 endpoint。

Microsoft 以逐行 GeoJSON 文件发布记录，并按国家和 Bing zoom level 9 quadkey 分区。本 Notebook 固定使用 **2026 年 7 月 24 日**版本的索引；该来源说明建筑轮廓来自 2014 至 2024 年间获取的影像。数据使用 [CDLA Permissive 2.0](https://cdla.dev/permissive-2-0/) 许可。Notebook 只下载 49 个地点所需的 Australia 分区，并把未经修改的源文件缓存到 `data/external_raw/`。

## 能从这个数据源得到什么

数据源直接提供的是建筑多边形 geometry。根据这些多边形，本 Notebook 可以计算指定范围内识别出的建筑轮廓数量、这些轮廓占用的地面面积，以及建筑轮廓占整个研究范围的比例。这些量可以作为周边开发强度的简单指标。多边形本身不提供建筑类型、用途、是否有人居住、入口、停车位、步行可达性或人口，因此本 Notebook 不会声称或推断这些信息。

## 本 Notebook 最终提取的 features

对于每个 `site_id`，Notebook 以地点坐标为中心建立精确的**直线距离 500 米缓冲区**，并生成以下独立 feature 表：

| 输出字段 | 直白含义 |
|---|---|
| `ms_building_count_500m` | 与 500 米缓冲区有实际面积重叠的建筑轮廓数量，表示地点周边识别出了多少个独立建筑 geometry。 |
| `ms_building_count_per_sqkm_500m` | 建筑轮廓数量除以缓冲区面积（平方公里），表示统一单位面积内的建筑出现频率。 |
| `ms_building_footprint_area_m2_500m` | 缓冲区内部被建筑轮廓覆盖的总地面面积，单位为平方米。缓冲区外的部分不计入，重叠的多边形不会重复计算。 |
| `ms_building_coverage_ratio_500m` | 建筑轮廓面积除以缓冲区总面积。例如 `0.10` 表示识别出的建筑轮廓覆盖了 500 米缓冲区的 10%。 |
| `ms_building_data_status` | 数据处理状态，用来区分成功且有建筑、成功但未检测到建筑，以及因为所需源数据分区不可用而无法完成。它是数据质量字段，不是环境特征。 |

当前所有地点都使用相同的 500 米缓冲区，因此建筑数量与单位面积数量包含相同的相对信息，建筑面积与覆盖率也包含相同的相对信息。建模时不应把每一对同时当作相互独立的预测变量。

## 1. 读取 `Model_Data` 并选择源数据分区

权威 `Model_Data` sheet 包含 49 个调查地点。本 Notebook 只读取 `site_id`、`latitude`、`longitude` 和 `coordinate_crs`；行人目标、`qa_status` 及其他项目字段不会复制到 feature 表。代码会验证 `site_id` 是非空且唯一的字符串，保证团队后续可以使用 `validate="one_to_one"` 合并。

当前使用**以地点为中心、直线距离 500 米的缓冲区**作为透明的测试参数。它不是客户已经确认的要求，也不是步行路网范围或沿道路走廊。每个地点在建立缓冲区前分别估计本地 UTM 投影，因此距离和面积不会直接在经纬度单位上计算。后续可以比较不同半径，或用经过验证的道路分段几何替换点，但数据源分区下载原则不需要改变。

本次使用的官方索引字段是：`Location` 表示来源地区；`QuadKey` 是 Bing zoom level 9 分区标识；`Url` 是分区下载地址；`Size` 是发布方给出的压缩文件大小标签；`UploadDate` 是源文件上传日期。`QuadKey` 是标识符，必须按文本读取。这里先用缓冲区外接矩形保守选择候选分区，之后再使用精确缓冲区计算相交结果。

```python
site_columns = ["site_id", "latitude", "longitude", "coordinate_crs"]
model_data = pd.read_excel(
    MODEL_DATASET_FILE,
    sheet_name="Model_Data",
    dtype={"site_id": "string"},
)
missing_site_columns = set(site_columns) - set(model_data.columns)
if missing_site_columns:
    raise ValueError(f"Model_Data fields are missing: {sorted(missing_site_columns)}")

sites = model_data[site_columns].copy()
sites["site_id"] = sites["site_id"].str.strip()
if len(sites) != 49:
    raise ValueError(f"Expected 49 surveyed sites, found {len(sites)}.")
if sites["site_id"].isna().any() or sites["site_id"].eq("").any():
    raise ValueError("site_id cannot be missing or blank.")
if sites["site_id"].duplicated().any():
    raise ValueError("site_id must be unique before feature extraction.")
if sites[["latitude", "longitude"]].isna().any().any():
    raise ValueError("All sites require latitude and longitude.")
if not sites["coordinate_crs"].eq("EPSG:4326").all():
    raise ValueError("All site coordinates must use WGS84 (EPSG:4326).")
if not sites["latitude"].between(-36, -13).all() or not sites["longitude"].between(112, 130).all():
    raise ValueError("One or more site coordinates fall outside the expected WA extent.")


def download_file(url: str, destination: Path) -> None:
    """Download once, using a temporary file so partial data are not cached."""
    if destination.is_file():
        return
    temporary = destination.with_name(destination.name + ".part")
    with requests.get(url, stream=True, timeout=(30, 300)) as response:
        response.raise_for_status()
        with temporary.open("wb") as output:
            for chunk in response.iter_content(chunk_size=1024 * 1024):
                if chunk:
                    output.write(chunk)
    temporary.replace(destination)


def tile_xy(latitude: float, longitude: float, zoom: int) -> tuple[int, int]:
    """Convert WGS84 coordinates to a Bing tile position."""
    latitude = min(max(latitude, -85.05112878), 85.05112878)
    scale = 1 << zoom
    x = (longitude + 180.0) / 360.0
    sin_latitude = math.sin(math.radians(latitude))
    y = 0.5 - math.log((1 + sin_latitude) / (1 - sin_latitude)) / (4 * math.pi)
    return (
        min(max(int(x * scale), 0), scale - 1),
        min(max(int(y * scale), 0), scale - 1),
    )


def quadkey(tile_x: int, tile_y: int, zoom: int) -> str:
    digits = []
    for level in range(zoom, 0, -1):
        mask = 1 << (level - 1)
        digit = (1 if tile_x & mask else 0) + (2 if tile_y & mask else 0)
        digits.append(str(digit))
    return "".join(digits)


def quadkeys_for_bounds(bounds: tuple[float, float, float, float], zoom: int) -> set[str]:
    west, south, east, north = bounds
    x_min, y_min = tile_xy(north, west, zoom)
    x_max, y_max = tile_xy(south, east, zoom)
    return {
        quadkey(x, y, zoom)
        for x in range(x_min, x_max + 1)
        for y in range(y_min, y_max + 1)
    }


analysis_context = {}
site_partition_rows = []
for row in sites.itertuples(index=False):
    point = gpd.GeoDataFrame(
        {"site_id": [row.site_id]},
        geometry=gpd.points_from_xy([row.longitude], [row.latitude]),
        crs="EPSG:4326",
    )
    metric_crs = point.estimate_utm_crs()
    buffer_metric = point.to_crs(metric_crs).geometry.iloc[0].buffer(
        BUFFER_RADIUS_M, resolution=32
    )
    buffer_wgs84 = gpd.GeoSeries([buffer_metric], crs=metric_crs).to_crs(4326).iloc[0]
    required_quadkeys = quadkeys_for_bounds(buffer_wgs84.bounds, QUADKEY_ZOOM)
    analysis_context[row.site_id] = {
        "metric_crs": metric_crs,
        "buffer_metric": buffer_metric,
        "buffer_wgs84": buffer_wgs84,
        "quadkeys": required_quadkeys,
    }
    site_partition_rows.extend(
        {"site_id": row.site_id, "QuadKey": key}
        for key in sorted(required_quadkeys)
    )

site_partitions = pd.DataFrame(site_partition_rows)
download_file(SOURCE_INDEX_URL, INDEX_FILE)
source_index = pd.read_csv(INDEX_FILE, dtype={"QuadKey": "string"})
required_index_columns = {"Location", "QuadKey", "Url", "Size", "UploadDate"}
if not required_index_columns.issubset(source_index.columns):
    raise ValueError("The pinned Microsoft index does not contain the expected fields.")

australia_index = source_index.loc[source_index["Location"].eq("Australia")].copy()
if australia_index["QuadKey"].duplicated().any():
    raise ValueError("The Australia index contains duplicate QuadKey values.")

required_quadkeys = set(site_partitions["QuadKey"])
selected_tiles = australia_index.loc[
    australia_index["QuadKey"].isin(required_quadkeys),
    ["QuadKey", "Url", "Size", "UploadDate"],
].sort_values("QuadKey")
missing_quadkeys = required_quadkeys - set(selected_tiles["QuadKey"])

display(sites)
display(selected_tiles.reset_index(drop=True))
print(f"Sites: {len(sites)}")
print(f"Required Australia tiles listed in the source index: {len(selected_tiles)}")
print(f"Required quadkeys not listed: {len(missing_quadkeys)}")
```

## 2. Download and parse only the selected building files

Each selected URL is an immutable source partition for the pinned release. The compressed filename ends in `.csv.gz`, but its contents are GeoJSON Features separated by newline, not a conventional comma-separated table. Each feature contains a polygon geometry and a `properties` object.

Only `geometry` is used in this pilot. The sampled Australia records also contain `height` and `confidence`, but observed values include `height = 0` and `confidence = -1`. Without a sufficiently clear field definition and a local missingness/validity assessment, treating those values as measured height or model probability would be unsafe. They are therefore deliberately excluded rather than converted into weak features.

The source files are preserved unchanged in the ignored raw-data directory. Parsing extracts geometry in memory; it does not rewrite the downloaded partitions. If a required quadkey is absent from the release index, the affected site is reported as unavailable instead of being assigned a false zero.

## 2. 只下载并解析选中的建筑文件

每个选中的 URL 都对应固定版本中的一个源数据分区。压缩文件名以 `.csv.gz` 结尾，但内容是每行一个 GeoJSON Feature，而不是普通的逗号分隔表。每条 feature 包含一个多边形 geometry 和一个 `properties` 对象。

本次测试只使用 `geometry`。抽查的 Australia 记录还包含 `height` 和 `confidence`，但实际可以看到 `height = 0` 和 `confidence = -1`。在缺少足够明确的字段定义及本地缺失和有效性分析时，把这些值当成实测高度或模型概率是不安全的，因此本阶段主动排除，而不是强行生成质量较弱的特征。

下载的源文件会原样保存在被 Git 忽略的原始数据目录中。解析只在内存中提取 geometry，不改写下载分区。如果某个所需 quadkey 没有出现在该版本索引中，受影响地点会标记为不可用，而不是被错误赋值为零。

```python
tile_paths = {}
for tile in selected_tiles.itertuples(index=False):
    tile_path = TILE_DIR / f"{tile.QuadKey}.csv.gz"
    download_file(tile.Url, tile_path)
    tile_paths[tile.QuadKey] = tile_path


def read_building_geometries(path: Path) -> gpd.GeoDataFrame:
    """Read geometry only from one Microsoft GeoJSONL gzip partition."""
    geometries = []
    with gzip.open(path, "rt", encoding="utf-8") as source:
        for line in source:
            if line.strip():
                geometries.append(shape(json.loads(line)["geometry"]))
    buildings = gpd.GeoDataFrame(geometry=geometries, crs="EPSG:4326")
    buildings = buildings.loc[
        buildings.geometry.notna() & ~buildings.geometry.is_empty
    ].copy()
    invalid = ~buildings.geometry.is_valid
    if invalid.any():
        buildings.loc[invalid, "geometry"] = buildings.loc[invalid, "geometry"].make_valid()
    return buildings


tile_cache = {key: read_building_geometries(path) for key, path in tile_paths.items()}
print(f"Downloaded/cached source tiles: {len(tile_cache)}")
print(f"Building footprint records loaded: {sum(len(data) for data in tile_cache.values()):,}")
```

## 3. Calculate a small, interpretable feature set

A source footprint is counted when its intersection with the exact 500 m buffer has positive area. Footprint area is calculated after clipping to the buffer; the union of clipped polygons is used so overlapping detections cannot double-count covered ground. All spatial measurements use the site's local UTM CRS.

| Output field | Definition and interpretation |
|---|---|
| `ms_building_count_500m` | Number of source footprint records with positive-area overlap with the buffer. It can be affected by merged or split detections. |
| `ms_building_count_per_sqkm_500m` | Count divided by the exact buffer area in square kilometres. This is the normalised building-frequency measure. |
| `ms_building_footprint_area_m2_500m` | Union area of footprint geometry inside the buffer, in square metres. Boundary buildings contribute only their in-buffer area. |
| `ms_building_coverage_ratio_500m` | Footprint union area divided by buffer area. This is the normalised built-cover measure; for example, `0.10` means 10% of the buffer is covered by detected footprints. |
| `ms_building_data_status` | Distinguishes a completed calculation with buildings, a completed calculation with no detected footprints, and a required partition missing from the release index. |

Raw counts and areas are retained for auditability, while the per-square-kilometre count and coverage ratio are the more comparable forms. Because every row currently uses the same 500 m radius, count and count density are exact rescalings of one another; footprint area and coverage ratio are also exact rescalings. They should not be entered together as separate predictors in the same model. The normalised fields become essential if later buffers differ in area.

## 3. 计算一组精简且可解释的特征

当源 footprint 与精确 500 米缓冲区的交集面积大于零时，才把它计入建筑数量。建筑面积先裁剪到缓冲区，再对裁剪后的多边形取并集，从而避免相互重叠的识别结果重复计算地面覆盖面积。所有空间测量都使用该地点的本地 UTM 坐标系。

| 输出字段 | 定义和解释 |
|---|---|
| `ms_building_count_500m` | 与缓冲区有正面积交集的源 footprint 记录数。识别结果的合并或拆分会影响这个数量。 |
| `ms_building_count_per_sqkm_500m` | 建筑数量除以精确缓冲区面积（平方公里），是标准化后的建筑出现频率。 |
| `ms_building_footprint_area_m2_500m` | 缓冲区内部 footprint 并集的面积，单位为平方米；跨越边界的建筑只计算缓冲区内部分。 |
| `ms_building_coverage_ratio_500m` | footprint 并集面积除以缓冲区面积，是标准化后的建筑覆盖指标；例如 `0.10` 表示缓冲区有 10% 被识别出的建筑轮廓覆盖。 |
| `ms_building_data_status` | 区分成功且检测到建筑、成功但未检测到 footprint，以及该版本索引缺少所需分区。 |

为便于审查，输出保留原始数量和面积；用于不同空间单位比较时，单位面积数量和覆盖率更合理。因为当前所有地点都采用相同的 500 米半径，数量与单位面积数量只是固定倍数关系，面积与覆盖率也是固定倍数关系，不应把每一对同时作为同一个模型中的独立预测变量。如果后续缓冲区面积不同，标准化字段才会变得不可替代。

```python
def summarise_site(site_id: str) -> dict:
    context = analysis_context[site_id]
    missing = context["quadkeys"] - set(tile_paths)
    if missing:
        return {
            "site_id": site_id,
            "ms_building_data_status": "source_tile_not_listed",
            "ms_building_count_500m": pd.NA,
            "ms_building_count_per_sqkm_500m": pd.NA,
            "ms_building_footprint_area_m2_500m": pd.NA,
            "ms_building_coverage_ratio_500m": pd.NA,
        }

    buildings = gpd.GeoDataFrame(
        pd.concat(
            [tile_cache[key] for key in sorted(context["quadkeys"])],
            ignore_index=True,
        ),
        geometry="geometry",
        crs="EPSG:4326",
    )
    candidates = buildings.loc[
        buildings.geometry.intersects(context["buffer_wgs84"])
    ].to_crs(context["metric_crs"])

    buffer_metric = context["buffer_metric"]
    buffer_area_m2 = buffer_metric.area
    intersections = candidates.geometry.intersection(buffer_metric)
    intersections = intersections.loc[
        ~intersections.is_empty & intersections.area.gt(0)
    ]

    building_count = len(intersections)
    footprint_area_m2 = unary_union(intersections.tolist()).area if building_count else 0.0
    status = (
        "complete_with_buildings"
        if building_count
        else "complete_no_detected_buildings"
    )
    return {
        "site_id": site_id,
        "ms_building_data_status": status,
        "ms_building_count_500m": building_count,
        "ms_building_count_per_sqkm_500m": building_count / (buffer_area_m2 / 1_000_000),
        "ms_building_footprint_area_m2_500m": footprint_area_m2,
        "ms_building_coverage_ratio_500m": footprint_area_m2 / buffer_area_m2,
    }


building_features = pd.DataFrame(
    summarise_site(site_id) for site_id in sites["site_id"]
)
building_features["ms_building_count_500m"] = (
    building_features["ms_building_count_500m"].astype("Int64")
)
for column, decimals in {
    "ms_building_count_per_sqkm_500m": 3,
    "ms_building_footprint_area_m2_500m": 3,
    "ms_building_coverage_ratio_500m": 6,
}.items():
    building_features[column] = pd.to_numeric(
        building_features[column], errors="coerce"
    ).round(decimals)

display(building_features)
```

## 4. 保存一张清晰的特征表

输出是独立的团队 feature 表，不是 `model_dataset.xlsx` 的副本。它保持每个 `site_id` 一行，只包含一个来源状态字段和四个建筑指标。坐标、道路字段、QA 字段和行人目标都不会复制，因为它们继续以 `Model_Data` 为权威来源。

代码在 `outputs/02_imagery_features/microsoft-building-footprints/` 下生成一个 `haoqi-07-microsoft-building-footprints-features.xlsx`。它只有 `Building_Features` 一个 sheet，因此 `pd.read_excel(path)` 可以直接读到 feature 表。之后建模代码按 `site_id` 使用 `how="left"` 和 `validate="one_to_one"` 合并。

```python
features = sites[["site_id"]].merge(
    building_features, on="site_id", how="left", validate="one_to_one"
)
output_columns = [
    "site_id",
    "ms_building_data_status",
    "ms_building_count_500m",
    "ms_building_count_per_sqkm_500m",
    "ms_building_footprint_area_m2_500m",
    "ms_building_coverage_ratio_500m",
]
features = features[output_columns]

if features["site_id"].duplicated().any() or len(features) != len(sites):
    raise ValueError("The feature table must contain one row per input site_id.")
if set(features["site_id"]) != set(sites["site_id"]):
    raise ValueError("Feature output site_id values do not match Model_Data.")

with pd.ExcelWriter(FEATURE_OUTPUT_FILE, engine="openpyxl") as writer:
    features.to_excel(writer, sheet_name="Building_Features", index=False)
    worksheet = writer.sheets["Building_Features"]
    worksheet.freeze_panes = "A2"
    worksheet.auto_filter.ref = worksheet.dimensions
    worksheet.sheet_view.showGridLines = False

    header_fill = PatternFill("solid", fgColor="1F4E78")
    for cell in worksheet[1]:
        cell.fill = header_fill
        cell.font = Font(color="FFFFFF", bold=True)
        cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
    worksheet.row_dimensions[1].height = 42

    for column, width in {"A": 14, "B": 31, "C": 27, "D": 36, "E": 39, "F": 36}.items():
        worksheet.column_dimensions[column].width = width
    for row in range(2, len(features) + 2):
        worksheet[f"C{row}"].number_format = "0"
        worksheet[f"D{row}"].number_format = "0.000"
        worksheet[f"E{row}"].number_format = "0.000"
        worksheet[f"F{row}"].number_format = "0.000000"
saved_features = pd.read_excel(
    FEATURE_OUTPUT_FILE, sheet_name="Building_Features", dtype={"site_id": "string"}
)
pd.testing.assert_frame_equal(
    saved_features,
    features.astype({"site_id": "string"}),
    check_dtype=False,
    check_exact=False,
    rtol=1e-9,
    atol=1e-9,
)

display(features)
display(features["ms_building_data_status"].value_counts(dropna=False).to_frame("sites"))
print("Saved:", FEATURE_OUTPUT_FILE.relative_to(PROJECT_ROOT))
```
